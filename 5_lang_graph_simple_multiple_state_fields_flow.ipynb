```python
#Import all importent depandancys
#LangGraph
# LangGraph
from langgraph.graph import StateGraph, START, END
# State management
from typing import TypedDict, Annotated
from IPython.display import Image, display
```

```python
from typing import TypedDict

class AgentState(TypedDict):
    name: str
    age: int
    city: str
    status: str
```

```python
def update_user(state: AgentState):
    return {
        "name": "Ajay Panchal",
        "age": 28
    }

def update_location(state: AgentState):
    return {
        "city": "Varanasi",
        "status": "Active"
    }
```

```python
graph = StateGraph(AgentState)
```

```python
graph.add_node("update_user", update_user)
graph.add_node("update_location", update_location)
```

```python
graph.add_edge(START, "update_user")
graph.add_edge("update_user", "update_location")
graph.add_edge("update_location", END)
```

```python
app = graph.compile()
#Print Graph
display(
    Image(app.get_graph().draw_mermaid_png())
)
```

```python

result = app.invoke({
    "name": "Ajay",
    "age": 27,
    "city": "Delhi",
    "status": "Inactive"
})

print(result)
```
