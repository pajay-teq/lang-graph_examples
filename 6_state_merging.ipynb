```python
#Import all importent depandancys
#LangGraph
# LangGraph
from langgraph.graph import StateGraph, START, END
# State management
from typing import TypedDict, Annotated
from IPython.display import Image, display
```

```python
from typing import TypedDict

class AgentState(TypedDict):
    name: str
    age: int
    city: str
    status: str
```

```python
def update_personal_info(state: AgentState):
    return {
        "name": "Ajay Panchal",
        "age": 28
    }

def update_location(state: AgentState):
    return {
        "city": "Varanasi",
        "status": "Active"
    }
```

```python
 

builder = StateGraph(AgentState)

builder.add_node("personal", update_personal_info)
builder.add_node("location", update_location)

builder.add_edge(START, "personal")
builder.add_edge("personal", "location")
builder.add_edge("location", END)

app = builder.compile() 
```

```python
#Print Graph
display(
    Image(app.get_graph().draw_mermaid_png())
)
```

```python


result = app.invoke({
    "name": "Ajay",
    "age": 27,
    "city": "Delhi",
    "status": "Inactive"
})

print(result)
```
