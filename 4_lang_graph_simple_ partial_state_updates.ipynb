```python
#Import all importent depandancys
#LangGraph
# LangGraph
from langgraph.graph import StateGraph, START, END
# State management
from typing import TypedDict, Annotated
from IPython.display import Image, display
```

```python
from typing import TypedDict

class AgentState(TypedDict):
    name: str
    age: int
    city: str
```

```python
def update_age(state: AgentState):
    return {
        "age": 28
    }
def update_city(state: AgentState):
    return {
        "city": "Delhi"
    }
```

```python
graph = StateGraph(AgentState)
```

```python
graph.add_node("update_age", update_age)
graph.add_node("update_city", update_city)
```

```python
graph.add_edge(START, "update_age")
graph.add_edge("update_age", "update_city")
graph.add_edge("update_city", END)
```

```python
app = graph.compile()
#Print Graph
display(
    Image(app.get_graph().draw_mermaid_png())
)
```

```python
result = app.invoke({
    "name": "Ajay",
    "age": 27,
    "city": "Varanasi"
})

print(result)
```
